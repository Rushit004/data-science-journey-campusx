###`Q1:` Join Tuples if similar initial element
While working with Python tuples, we can have a problem in which we need to perform concatenation of records from the similarity of initial element. This problem can have applications in data domains such as Data Science.

For eg.
```
Input  : test_list = [(5, 6), (5, 7), (5, 8), (6, 10), (7, 13)]
Output : [(5, 6, 7, 8), (6, 10), (7, 13)]
```

````python
# copy paste
test_list = [(5, 6), (5, 7), (5, 8), (6, 10), (7, 13)]


res_dict = {}
for key, *values in test_list:
    if key in res_dict:
        res_dict[key].extend(values)
    else:
        res_dict[key] = [key] + list(values)


result = [tuple(val) for val in res_dict.values()]

print(result)






````

```
The original tuple : (1, 5, 7, 8, 10)
Resultant tuple after multiplication :

(1*5, 1*5+5*7, 7*5 + 7*8, 8*7 + 8*10, 10*8) -> (5, 40, 91, 136, 80)

output-(5, 40, 91, 136, 80)
```

````python
# write your code here
t = (1, 5, 7, 8, 10,11)
x = len(t)-1

l = [t[0]*t[1]]
for i in range(1,len(t)-1):
  l.append(t[i]*t[i+1] + t[i]*t[i-1])

l.append(t[x-1]*t[x])

print(tuple(l))

````

###`Q3`: Check is tuples are same or not?
Two tuples would be same if both tuples have same element at same index
```
t1 = (1,2,3,0)
t2 = (0,1,2,3)

t1 and t2 are not same
```

###`Q4`: Count no of tuples, list and set from a list
```
list1 = [{'hi', 'bye'},{'Geeks', 'forGeeks'},('a', 'b'),['hi', 'bye'],['a', 'b']]

```
`Output:`

```
List-2
Set-2
Tuples-1
```

````python
# write your code here
list1 = [{'hi', 'bye'},{'Geeks', 'forGeeks'},('a', 'b'),['hi', 'bye'],['a','b']]
setcnt = 0
tuplecnt = 0
listcnt = 0
for i in list1:
  if type(i)==set:
    setcnt+=1
  elif type(i)==tuple:
    tuplecnt+=1
  elif type(i)==list:
    listcnt+=1
  else:
    print("Invalid type for count")

print("Lists",listcnt,sep = '-')
print("Sets",setcnt,sep = '-')
print("Tuples",tuplecnt,sep = '-')





````

###`Q1:` Write a program to find set of common elements in three lists using sets.
```
Input : ar1 = [1, 5, 10, 20, 40, 80]
        ar2 = [6, 7, 20, 80, 100]
        ar3 = [3, 4, 15, 20, 30, 70, 80, 120]

Output : [80, 20]
```

````python
# write your code here
ar1 = [1, 5, 10, 20, 40, 80]
ar2 = [6, 7, 20, 80, 100]
ar3 = [3, 4, 15, 20, 30, 70, 80, 120]

s1 = set(ar1)
s2 = set(ar2)
s3 = set(ar3)

print(list(s1.intersection(s2.intersection(s3))))
````

### `Q5`: Sort Dictionary key and values List.

**Example 1:**

Input:

```bash
{'c': [3], 'b': [12, 10], 'a': [19, 4]}
```

Output:

```bash
{'a': [4, 19], 'b': [10, 12], 'c': [3]}
```

**Example 2:**

Input:

```bash
{'c': [10, 34, 3]}
```

Output:

```bash
{'c': [3, 10, 34]}
```

````python
# hint needed
# write your code here

d1 = {'c': [3], 'b': [12, 10], 'a': [19, 4]}

for i in d1:
  d1[i].sort()


sorted_keys = sorted(d1.keys())

result = {}

for i in sorted_keys:
  result[i] = d1[i]
print(result)

````
