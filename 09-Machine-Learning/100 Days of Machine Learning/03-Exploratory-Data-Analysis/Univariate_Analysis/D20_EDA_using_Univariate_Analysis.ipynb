```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('Datasets/D20/train.csv')
```

```python
df.head()
```

# 1.Categorical Data

## a. CountPlot

```python
sns.countplot(data=df,x = 'Survived')
plt.show()
```

```python
df['Pclass'].value_counts().plot(kind = 'bar')
```

## b. Pie Chart

```python
df['Survived'].value_counts().plot(kind = 'pie',autopct = '%.2f')
```

# 2. Numerical Data

## a. Histogram

```python
plt.hist(df['Age'],bins = 10)
```

## b. Distplot

```python
sns.distplot(df['Age'])
```

```python
sns.histplot(data = df,x = 'Age',kde = True,stat="density")
```

```python
sns.kdeplot(data = df,x = 'Age')
```

## c. Box plot

```python
sns.boxplot(df['Fare'])
```

```python
sns.boxplot(df['Age'])
```

```python
df['Age'].skew()
```
