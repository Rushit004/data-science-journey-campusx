```python
iris = load_iris()

X_train = iris.data
y_train = iris.target

clas = DecisionTreeClassifier(random_state=42)
clas.fit(X_train, y_train)

X = iris.data[np.random.randint(0, len(iris.data))]

print("Selected sample:")
print(X)
```

```python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt

plt.figure(figsize=(16, 9))

plot_tree(
    clas,
    feature_names=iris.feature_names,
    class_names=iris.target_names,
    filled=True,
    rounded=True,
    node_ids=True
)

plt.show()
```

```python
plt.figure(figsize=(16, 8))

plot_tree(
    clas,
    feature_names=iris.feature_names,
    class_names=iris.target_names,
    filled=True,
    node_ids=True
)

plt.show()
```

```python
from sklearn.tree import export_text

tree_rules = export_text(
    clas,
    feature_names=iris.feature_names
)

print(tree_rules)
```

```python
X = iris.data[np.random.randint(0, len(iris.data))]

print("Input:")
print(X)

print("\nPredicted class:")
print(iris.target_names[clas.predict([X])[0]])
```

```python
node_indicator = clas.decision_path([X])

leaf_id = clas.apply([X])

node_ids = node_indicator.indices[
    node_indicator.indptr[0]:node_indicator.indptr[1]
]

print("Nodes followed by the sample:")
print(node_ids)

print("Leaf node:")
print(leaf_id[0])
```

```python
node_indicator = clas.decision_path([X])

node_ids = node_indicator.indices[
    node_indicator.indptr[0]:node_indicator.indptr[1]
]

print("Prediction path:\n")

for node_id in node_ids:
    if clas.tree_.children_left[node_id] == clas.tree_.children_right[node_id]:
        print(f"Reached leaf node {node_id}")
        continue

    feature_id = clas.tree_.feature[node_id]
    threshold = clas.tree_.threshold[node_id]

    if X[feature_id] <= threshold:
        direction = "<="
    else:
        direction = ">"

    print(
        f"{iris.feature_names[feature_id]} "
        f"{direction} {threshold:.2f}"
    )
```

```python
feature_importance = pd.DataFrame({
    "Feature": iris.feature_names,
    "Importance": clas.feature_importances_
})

feature_importance = feature_importance.sort_values(
    "Importance",
    ascending=False
)

print(feature_importance)
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    feature_importance["Feature"],
    feature_importance["Importance"]
)

plt.xticks(rotation=45)
plt.ylabel("Importance")
plt.title("Decision Tree Feature Importance")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.tree import DecisionTreeRegressor

df_cars = pd.read_csv("DataSets/D83/cars.csv")

X = df_cars[["WGT"]]
y = df_cars["MPG"]

dt = DecisionTreeRegressor(
    max_depth=3,
    criterion="absolute_error",
    random_state=42
)

dt.fit(X, y)
```

```python
X_sorted = X.sort_values("WGT")

y_pred = dt.predict(X_sorted)

plt.figure(figsize=(8, 5))

plt.scatter(
    X["WGT"],
    y,
    label="Actual"
)

plt.plot(
    X_sorted["WGT"],
    y_pred,
    linewidth=2,
    label="Decision Tree"
)

plt.xlabel("WGT")
plt.ylabel("MPG")
plt.title("Decision Tree Univariate Regression")
plt.legend()

plt.show()
```
