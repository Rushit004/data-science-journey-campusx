```python
import pandas as pd
import requests
```

```python
respons = requests.get('https://api.themoviedb.org/3/movie/top_rated?api_key=4e4c212e92d8641e313cf538122fb21f&language=en-US&page=1')
```

```python
df = pd.DataFrame(respons.json()['results'])[['id','title','overview','release_date','popularity','vote_average','vote_count']]
```

```python
df
```

```python
df = pd.DataFrame()
```

```python
df
```

```python
for i in range(1,429):
    respons = requests.get('https://api.themoviedb.org/3/movie/top_rated?api_key=4e4c212e92d8641e313cf538122fb21f&language=en-US&page={}'.format(i))
    temp_df = pd.DataFrame(respons.json()['results'])[['id','title','overview','release_date','popularity','vote_average','vote_count']]
    df = pd.concat([df, temp_df], ignore_index=True)
```

```python
df
```

```python
df.to_csv('movies.csv')
```
