```python
name = []
rating = []
review_count = []
about = []
salaries = []
interviews = []
benefits = []

for i in company:
    name.append(i.find('h2', class_='companyCardWrapper__companyName').text.strip())
    rating.append(i.find('div', class_='rating_star_container').text.strip())
    review_count.append(i.find('span', class_='companyCardWrapper__companyRatingCount').text.strip())
    about.append(i.find('span', class_='companyCardWrapper__interLinking').text.strip().split('|')[0].strip())

    # tertiary info block (salaries, interviews, benefits, etc.)
    tertiary_info = i.find('div', class_='companyCardWrapper__tertiaryInformation')
    
    data = {}
    if tertiary_info:
        actions = tertiary_info.find_all('a', class_='companyCardWrapper__ActionWrapper')
        for action in actions:
            count = action.find('span', class_='companyCardWrapper__ActionCount')
            title = action.find('span', class_='companyCardWrapper__ActionTitle')
            if count and title:
                data[title.get_text(strip=True)] = count.get_text(strip=True)

    salaries.append(data.get('Salaries'))
    interviews.append(data.get('Interviews'))
    benefits.append(data.get('Benefits'))
```

```python
data = {'company_name':name,"sector ":about,"Rating":rating,"reviews":review_count,"Average salary":salaries,"Number of Interview":interviews,"Benifited":benefits}
```

```python
company_data = pd.DataFrame()

for j in range(1,100):
    url = 'https://www.ambitionbox.com/list-of-companies?campaign=desktop_nav&page={}'.format(j)
    
    headers={'User-Agent':'Mozilla/5.0 (Windows NT 6.3; Win 64 ; x64) Apple WeKit /537.36(KHTML , like Gecko) Chrome/80.0.3987.162 Safari/537.36'}
    webpage = requests.get(url,headers = headers).text

    soup = BeautifulSoup(webpage,'lxml')
    company = soup.find_all('div',itemtype='http://schema.org/ListItem')
    name = []
    rating = []
    review_count = []
    about = []
    salaries = []
    interviews = []
    benefits = []
    
    for i in company:
        name.append(i.find('h2', class_='companyCardWrapper__companyName').text.strip())
        rating.append(i.find('div', class_='rating_star_container').text.strip())
        review_count.append(i.find('span', class_='companyCardWrapper__companyRatingCount').text.strip())
        about.append(i.find('span', class_='companyCardWrapper__interLinking').text.strip().split('|')[0].strip())
    
        # tertiary info block (salaries, interviews, benefits, etc.)
        tertiary_info = i.find('div', class_='companyCardWrapper__tertiaryInformation')
        
        data = {}
        if tertiary_info:
            actions = tertiary_info.find_all('a', class_='companyCardWrapper__ActionWrapper')
            for action in actions:
                count = action.find('span', class_='companyCardWrapper__ActionCount')
                title = action.find('span', class_='companyCardWrapper__ActionTitle')
                if count and title:
                    data[title.get_text(strip=True)] = count.get_text(strip=True)
    
        salaries.append(data.get('Salaries'))
        interviews.append(data.get('Interviews'))
        benefits.append(data.get('Benefits'))

        data = {'company_name':name,"sector ":about,"Rating":rating,"reviews":review_count,"Average salary":salaries,"Number of Interview":interviews,"Benifited":benefits}
        df = pd.DataFrame(data)

        company_data = pd.concat([company_data, df], ignore_index=True)
```
