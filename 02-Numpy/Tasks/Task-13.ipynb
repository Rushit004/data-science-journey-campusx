Create a python function named **date_array** that accepts two dates as string format and returns a numpy array of dates between those 2 dates. The function only accept 2 strings, otherwise raise error. The date format should be like this only: `2022-12-6`. The end date should be included and for simplicity, choose dates from a same year.



**Example 1:**

```bash
date_array(start = '2020-09-15', end = '2020-09-25')
```

Output:

```bash
['2020-09-15', '2020-09-16', '2020-09-17', '2020-09-18',
```

**Example 3:**


Output:
```bash
['2020-11-25', '2020-11-26', '2020-11-27', '2020-11-28',
 '2020-11-29', '2020-11-30']
```

````python
#Code here
# def create_Random_dates(sdate,edate):
#   s = np.array(sdate.split('-'))
#   e = np.array(edate.split('-'))

#   print(s)
#   print(e)



# sdate = input('Enter starting date.')
# edate = input('Enter Ending date.')
# create_Random_dates(sdate,edate)

# copied
import numpy as np
from datetime import datetime

def date_array(start, end):
    # Validate exactly 2 string arguments
    if not isinstance(start, str) or not isinstance(end, str):
        raise TypeError("Both arguments must be strings.")

    # Validate date format
    date_format = "%Y-%m-%d"
    try:
        start_date = datetime.strptime(start, date_format)
        end_date   = datetime.strptime(end, date_format)
    except ValueError:
        raise ValueError("Date format must be 'YYYY-MM-DD', e.g. '2022-12-6'.")

    # Validate start is before end
    if start_date > end_date:
        raise ValueError("Start date must be before or equal to end date.")

    # Generate date range (inclusive of end)
    dates = np.arange(start, end, dtype="datetime64[D]")
    dates = np.append(dates, np.datetime64(end, "D"))  # include end date

    return dates.astype(str)


# Example
print(date_array(start="2020-09-15", end="2020-09-25"))


````

###`Q-19:` Write a function which will accept 2 arguments.
First: A 1D numpy array arr

Second: An integer n {Please make sure n<=len(arr)}

Output: The output should be the nth largest item out of the array
```
# Example1 : arr=(12,34,40,7,1,0) and n=3, the output should be 12
# Example2 : arr=(12,34,40,7,1,0) and n=1, the output should be 40
```

````python
#Code here
def nth_largest(arr,n):
  if n>len(arr):
    print('Error.')
  else:
    arr.sort()
    print(arr[-n])

arr = np.array([2,4,5,8,3,4,0])
nth_largest(arr,2)
````
