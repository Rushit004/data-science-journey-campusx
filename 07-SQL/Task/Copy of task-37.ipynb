###`Q-1`: Rank Employee in terms of revenue generation. Show employee id, first name, revenue, and rank


###`Q-2`: Show All products cumulative sum of units sold each month.


###`Q-3`: Show Percentage of total revenue by each suppliers



###`Q-4:` Show Percentage of total orders by each suppliers



### `Q-5`:Show All Products Year Wise report of totalQuantity sold, percentage change from last year.





````python
# 1.Rank Employee in terms of revenue generation. Show employee id, first name, revenue, and rank

SELECT t3.EmployeeID ,t3.FirstName, SUM(t2.Quantity*t2.UnitPrice) AS 'revenue',
DENSE_RANK() OVER(ORDER BY SUM(t2.Quantity*t2.UnitPrice) DESC) AS 'rank'
FROM northwind.nw_orders t1
JOIN northwind.nw_order_details t2
ON t1.OrderID = t2.OrderID
JOIN northwind.nw_employees t3
ON t1.EmployeeID = t3.EmployeeID
GROUP BY EmployeeID,FirstName

# 2.Show All products cumulative sum of units sold each month.
SELECT MONTHNAME(t3.OrderDate),t2.ProductID,t2.ProductName, SUM(t1.Quantity) AS 'total',
SUM(SUM(t1.Quantity)) OVER(ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW) AS 'Cum_sum'
FROM northwind.nw_order_details t1
JOIN northwind.nw_products t2
ON t1.ProductID = t2.ProductID
JOIN northwind.nw_orders t3
ON t1.OrderID = t3.OrderID
GROUP BY MONTHNAME(t3.OrderDate),t2.ProductID,t2.ProductName

# 3.Show Percentage of total revenue by each suppliers
SELECT t.SupplierID,( revenue / SUM(revenue) OVER())*100 AS 'percentage'
FROM(
SELECT t2.SupplierID,SUM(t3.Quantity*t3.UnitPrice) AS 'revenue'
FROM northwind.nw_products t1
JOIN northwind.nw_suppliers t2
ON t1.SupplierID = t2.SupplierID
JOIN northwind.nw_order_details t3
ON t1.ProductID = t3.ProductID
GROUP BY t2.SupplierID
) t
ORDER BY percentage DESC

# 4. Show Percentage of total orders by each suppliers

SELECT t.SupplierID,( qty_orders / SUM(qty_orders) OVER())*100 AS 'percentage'
FROM(
SELECT t2.SupplierID,SUM(t3.Quantity) AS 'qty_orders'
FROM northwind.nw_products t1
JOIN northwind.nw_suppliers t2
ON t1.SupplierID = t2.SupplierID
JOIN northwind.nw_order_details t3
ON t1.ProductID = t3.ProductID
GROUP BY t2.SupplierID
) t
ORDER BY percentage DESC

# 5.Show All Products Year Wise report of totalQuantity sold, percentage change from last year.
SELECT YEAR(t1.OrderDate) AS yr,SUM(t2.Quantity) AS total_qty_sold,
100 * (SUM(t2.Quantity) - LAG(SUM(t2.Quantity)) OVER (ORDER BY YEAR(t1.OrderDate)))
    / LAG(SUM(t2.Quantity)) OVER (ORDER BY YEAR(t1.OrderDate)) AS percentage
FROM northwind.nw_orders t1
JOIN northwind.nw_order_details t2
    ON t1.OrderID = t2.OrderID
GROUP BY YEAR(t1.OrderDate);
````
