### Annotations

```python
batter = pd.read_csv('Dataset/batter.csv')
batter.head()
```

```python
batter.shape
```

```python
sample_df = batter.head(100).sample(25,random_state = 5)
sample_df
```

```python
plt.scatter(sample_df['avg'],sample_df['strike_rate'])
```

```python
x = [1,2,3,4]
y = [5,6,7,8]

plt.scatter(x,y)
plt.text(1,5,'Pt1')
plt.text(2,6,'Pt2')
plt.text(3,7,'Pt3',fontdict = {'size':12,'color' :'red'})
plt.text(4,8,'Pt4')
```

```python
plt.figure(figsize = (18,10))
plt.scatter(sample_df['avg'],sample_df['strike_rate'],s = sample_df['runs'])
for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])

plt.xlabel('Average')
plt.ylabel('Strike rate')
```

### Horizontal and Vertical lines

```python

plt.figure(figsize = (18,10))
plt.scatter(sample_df['avg'],sample_df['strike_rate'],s = sample_df['runs'])

plt.axhline(130,color = 'red')
plt.axhline(140,color = 'green')
plt.axvline(30,color = 'red')

for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])

plt.xlabel('Average')
plt.ylabel('Strike rate')
```

###  3D Surface Plot

```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
```

```python
xx,yy = np.meshgrid(x,y)
xx.shape
```

```python
z = xx*xx + yy*yy
z.shape
```

```python
fig = plt.figure(figsize = (8,8))

ax = plt.subplot(projection = '3d')

g = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(g)
```

```python
z = np.sin(xx) + np.cos(yy)
fig = plt.figure(figsize = (8,8))

ax = plt.subplot(projection = '3d')

g = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(g)
```

### Contour Plots

```python
z = xx*xx + yy*yy

fig = plt.figure(figsize = (8,8))

ax = plt.subplot()

g = ax.contour(xx,yy,z,cmap='viridis')
fig.colorbar(g)
```

```python
z = xx*xx + yy*yy

fig = plt.figure(figsize = (8,8))

ax = plt.subplot()

g = ax.contourf(xx,yy,z,cmap='viridis')
fig.colorbar(g)
```

```python
z = np.sin(xx) + np.cos(yy)

fig = plt.figure(figsize=(12,8))

ax = plt.subplot()

p = ax.contourf(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

### Heatmap

```python
delivery = pd.read_csv('Dataset/IPL_Ball_by_Ball_2008_2022.csv')
delivery.head()
```

```python
temp_df = delivery[(delivery['ballnumber'].isin([1,2,3,4,5,6])) & (delivery['batsman_run'] == 6)]
temp_df.head()
```

```python
grid = temp_df.pivot_table(index = 'overs',columns = 'ballnumber',values = 'batsman_run',aggfunc = 'count')
grid
```

```python
plt.figure(figsize =(10,8)) 

plt.yticks(delivery['overs'].unique(), list(range(1,21)))
plt.xticks(np.arange(0,6), list(range(1,7)))

plt.imshow(grid)
plt.colorbar()
plt.title('6 hited ' )
plt.xlabel('Ball')
plt.ylabel('Over')
```
