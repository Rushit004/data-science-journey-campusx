### 2D Line plot

image.png

- Bivariate Analysis
- categorical -> numerical and numerical -> numerical
- Use case - Time series data

```python

price = [48000,54000,57000,49000,47000,45000]
year = [2015,2016,2017,2018,2019,2020]

plt.plot(year,price)
```

```python
# from a pandas dataframe
batsman = pd.read_csv("Dataset/sharma-kohli.csv")
batsman
```

```python
plt.plot(batsman['index'],batsman['V Kohli'])
```

```python
# plotting multiple plots

plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
```

```python
# labels title
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])

plt.title('Rohit sharma vs Virat Kohli runs comparision')
plt.xlabel('Season')
plt.ylabel('Runs')
```

```python
# colors(hex) and line(width and style) and marker(size)

plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle = 'dashed',linewidth = 2)
plt.plot(batsman['index'],batsman['RG Sharma'],color='blue',linestyle = 'dashdot',marker = '*',markersize = 10)

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
# legend -> location
plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle = 'dashed',linewidth = 2,label = 'virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color='blue',linestyle = 'dashdot',marker = '*',markersize = 10,label = 'rohit')

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')

plt.legend(loc = 'best')
```

```python
# limiting axes
price = [48000,54000,57000,49000,47000,45000,4500000]
year = [2015,2016,2017,2018,2019,2020,2021]

plt.plot(year,price)

plt.ylim(0,75000)
plt.grid()
```

```python
# show
plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F',linestyle='solid',linewidth=3,marker='D',markersize=10)
plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6',linestyle='dashdot',linewidth=2,marker='o')

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')

plt.grid()

plt.show()
```

```python
# plt.scatter on pandas data

df = pd.read_csv('Dataset/batter.csv')
df = df.head(50)
```

```python
plt.scatter(df['avg'],df['strike_rate'],color = 'red',marker = '+')

plt.title('AVG vs Strike rate analysis of top 50 batsman')
plt.xlabel('AVG of batsman')
plt.ylabel('Strike Rate of batsman')
```

```python
# size
tips = sns.load_dataset('tips')
tips
```

```python
plt.scatter(tips['total_bill'],tips['tip'])
```

```python
# slower
plt.scatter(tips['total_bill'],tips['tip'],s = tips['size']*20)
```

```python
# faster compared to plt.scatter
# scatterplot using plt.plot

plt.plot(tips['total_bill'],tips['tip'],'o')
```

*italicized text*### Bar chart

image.png

- Bivariate Analysis
- Numerical vs Categorical
- Use case - Aggregate analysis of groups

```python
# simple bar chart
children = [10,20,40,10,30]
colors = ['red','blue','green','yellow','pink']

plt.bar(colors,children,color = 'red')
plt.title('Children loves Color')
plt.xlabel('Color')
plt.ylabel('Number of childrens')
```

```python
df = pd.read_csv('Dataset/batsman_season_record.csv')
df
```

```python
plt.bar(df['batsman'],df['2015'],width = 0.35)
```

```python
plt.bar(np.arange(0,5)-0.2,df['2015'],width = 0.2)
plt.bar(np.arange(0,5),df['2016'],width = 0.2)
plt.bar(np.arange(0,5)+0.2,df['2017'],width = 0.2)

plt.xticks(np.arange(df.shape[0]),df['batsman'])
plt.title('Batsman Runs Season wise')
plt.xlabel('Batsman')
plt.ylabel('Runs')
plt.show()
```

```python
# a problem
children = [10,20,40,10,30]
colors = ['red red red red red red','blue blue blue blue','green green green green green','yellow yellow yellow yellow ','pink pinkpinkpink']

plt.bar(colors,children,color='black')
plt.xticks(rotation= 10)
```

```python
# Stacked Bar chart

plt.bar(df['batsman'],df['2017'],label='2017')
plt.bar(df['batsman'],df['2016'],bottom=df['2017'],label='2016')
plt.bar(df['batsman'],df['2015'],bottom=(df['2016'] + df['2017']),label='2015')

plt.legend()
plt.show()
```

### Pie Chart

image.png

- Univariate/Bivariate Analysis
- Categorical vs numerical
- Use case - To find contibution on a standard scale

```python
# simple data
data = [23,45,100,20,49]
subjects = ['eng','science','maths','sst','hindi']
plt.pie(data,labels=subjects)

plt.show()
```

```python
df = pd.read_csv('Dataset/gayle-175.csv')
df
```

```python
plt.pie(df['batsman_runs'],labels = df['batsman'],autopct = '%0.01f%%')
plt.show()
```

```python
plt.pie(df['batsman_runs'],labels = df['batsman'],autopct = '%0.01f%%',explode = [0.2,0,0,0,0,0],shadow = True)
plt.show()
```
